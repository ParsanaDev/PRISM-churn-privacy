# Telecom Churn Dataset analysis

```python
import numpy as np
import pandas as pd
```

```python
# Load the data from csv file into a pandas DataFrame
df = pd.read_csv('../../data/raw/Telco_customer_churn.csv')
```

```python
# Columns
df.columns
```

```python
# Quick statistical summary of the data
df.describe()
```

```python
# Customers who churned (Churn Value == 1), with detailed columns
detail_cols = ['CustomerID', 'Tenure Months', 'Phone Service', 'Churn Reason',
               'Internet Service', 'Online Security', 'Online Backup', 'Device Protection',
               'Tech Support', 'Streaming TV', 'Streaming Movies', 'Contract',
               'Paperless Billing', 'Payment Method', 'Monthly Charges', 'Total Charges']
churned = df.loc[df['Churn Value'] == 1, detail_cols]
churned
```

```python
# Unique churn reasons and how often each occurs (separate columns)
reason_counts = (churned['Churn Reason']
                 .value_counts()
                 .rename_axis('Churn Reason')
                 .reset_index(name='Count'))
reason_counts
```
